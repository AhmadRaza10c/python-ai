```python
"""Lesson 1.7: Pick the right AI model for a task, like an agent choosing a tool."""

task = input("Task (code / private / chat): ").strip().lower()
tokens = int(input("Estimated tokens: "))

# Private data must never leave the laptop, so check it first
if task == "private":
    model = "llama3.2 (local, Ollama)"
elif task == "code":
    model = "claude (cloud, strong at code)"
else:
    model = "small cloud model (cheap for chat)"

# Very long requests cost more, so warn before sending
if tokens > 4000:
    warning = "Long request: consider shortening it."
else:
    warning = "Length is fine."

print("=" * 40)
print(f"Chosen model: {model}")
print(warning)
```

```python
"""Lesson 1.6: Prompt Cleaner - cleans, measures and safely sizes a customer question."""

raw_name = input("Your name: ")
raw_question = input("Your question: ")

# Why: strip() removes extra spaces and title()/capitalize() fix random capitals,
# so the AI always receives a tidy name and question
clean_name = raw_name.strip().title()
clean_question = raw_question.strip().capitalize()

# Stretch: hide a bad word before the prompt is built
clean_question = clean_question.replace("stupid", "***")

characters = len(clean_question)

# Why: about 4 characters = 1 token, so this is a quick size estimate before paying an AI
estimated_tokens = round(characters / 4)

# Why: lower() first, so "Python", "PYTHON" and "python" are all found
mentions_python = "python" in clean_question.lower()

# Why: slicing to 150 characters stops one huge question from becoming an expensive prompt
short_question = clean_question[:150]

prompt = f"""You are a friendly Python tutor.
Greet {clean_name} by name and answer in 3 short lines.
Question: {short_question}"""

# Stretch: estimated cost with 2 decimals (example price 0.5 PKR per 1K tokens)
estimated_cost = estimated_tokens / 1000 * 0.5

print("=" * 40)
print(f"Cleaned question: {clean_question}")
print(f"Characters: {characters}, about {estimated_tokens} tokens")
print(f"Mentions Python: {mentions_python}")
print(f"Estimated cost: {estimated_cost:.2f} PKR")
print("=" * 40)
print(prompt)
```

```python
"""Lesson 1.7: Smart Model Router - sends each task to the cheapest model that is good enough."""

# Why: strip().lower() so " Code " and "CODE" still work
task = input("Task (code / private / chat): ").strip().lower()
tokens = int(input("Estimated tokens: "))
paid_answer = input("Is this a paid user? (yes/no): ").strip().lower()

# Why: private data must stay on the computer, so it goes to the local model with price 0
if task == "code":
    model = "claude (cloud, strong at code)"
    price_per_1k = 0.8  # example price in PKR
elif task == "private":
    model = "llama3.2 (local, Ollama)"
    price_per_1k = 0
else:
    model = "small cloud model (cheap)"
    price_per_1k = 0.3

# Why: a long request costs more, so we warn the user before it is sent
if tokens > 4000:
    print("Warning: long request, consider shortening it.")
else:
    print("Length is fine.")

# Stretch: free users cannot send long requests (uses "and")
if paid_answer == "no" and tokens > 2000:
    print("Upgrade to send long requests.")

cost = tokens / 1000 * price_per_1k
print(f"Chosen model: {model}")
print(f"Estimated cost: {cost:.2f} PKR")
```

```python
"""Lesson 1.8: Rule-based chatbot for Sweet Crumbs bakery (example business)."""

BOT = "Sweet Crumbs Bot"
print(f"{BOT}: Hi! Ask about hours, prices, location or orders. Type 'bye' to exit.")

# Why: the name is asked ONCE before the loop, so the bot remembers it for the whole chat
name = input(f"{BOT}: What is your name? ").strip().title()
if name == "":
    name = "Friend"

count = 0
unanswered = 0
complaints = 0

while True:
    message = input(f"{name}: ").strip().lower()
    count += 1

    # Why: spaces around the text let us match whole words only,
    # so " late " is found in "my order is late" but NOT inside "chocolate"
    clean_text = message.replace("?", "").replace(".", "").replace(",", "").replace("!", "")
    padded = " " + clean_text + " "

    if message == "bye":
        print(f"{BOT}: Take care, {name}!")
        break
    elif message == "":
        print(f"{BOT}: Please type a question.")
    # Why: handover and complaints come first, so an unhappy customer always reaches a human
    elif message == "agent" or message == "human":
        print(f"{BOT}: Connecting you to a human...")
        break
    elif " refund " in padded or " complaint " in padded or " late " in padded:
        complaints += 1
        print(f"{BOT}: Sorry about that. A manager will call you within 1 hour.")
    elif "hour" in message or "open" in message or "timing" in message or " time " in padded:
        print(f"{BOT}: We are open 9 am to 10 pm, every day.")
    elif "price" in message or "cost" in message or "how much" in message:
        print(f"{BOT}: Cakes start at 1,500 PKR. Cupcakes are 250 PKR each.")
    elif "where" in message or "location" in message or "address" in message:
        print(f"{BOT}: We are on Main Boulevard, Lahore.")
    elif "order" in message or "delivery" in message:
        print(f"{BOT}: Order on WhatsApp. Delivery is free over 3,000 PKR.")
    # Why: short words like "hi" are checked with == / startswith, because "hi" hides inside "which"
    elif message == "hi" or message == "hello" or message.startswith("salam"):
        print(f"{BOT}: Hello {name}! How can I help?")
    else:
        unanswered += 1
        question = message.capitalize()[:150]
        print(f"{BOT}: I am not sure yet. In Module 2, an AI will answer: "
              f"You are the friendly assistant of Sweet Crumbs bakery. "
              f"Answer in 2 lines: {question}")

print("=" * 40)
print(f"Chat summary: {count} messages, {unanswered} unanswered.")
print(f"Complaints received: {complaints}")
```

```python
"""AI Front Desk for Sweet Crumbs bakery (example business).

Answers common questions with free rules, gives exact price quotes (discount + delivery),
and turns every question it cannot answer into a cost-checked AI prompt.
Uses Module 1 tools only: no lists, dictionaries, for loops or functions.
"""

# Why: constants sit at the top, so the owner can change a price or fee in one place
BOT_NAME = "Sweet Crumbs Desk"
BUSINESS = "Sweet Crumbs bakery"
CUPCAKE_PRICE = 250
BROWNIE_PRICE = 300
CAKE_PRICE = 1500
BULK_MIN_QTY = 10
BULK_DISCOUNT = 0.10
DELIVERY_FEE = 200
FREE_DELIVERY_LIMIT = 3000
CLOUD_PRICE_PER_1K = 0.1  # example price in PKR per 1K tokens, not a real quote

print("=" * 50)
print("   Sweet Crumbs Desk: rules, quotes and AI-ready prompts")
print("=" * 50)
print("Ask anything, type /quote for a price, or bye to exit.")

# Why: the name is asked once, before the loop, so the desk remembers it for the whole chat
name = input(f"{BOT_NAME}: What is your name? ").strip().title()
if name == "":
    name = "Friend"

count = 0
answered = 0
unanswered = 0
quotes = 0
quoted_value = 0
complaints = 0
ai_cost = 0

while True:
    message = input(f"{name}: ").strip().lower()
    if message != "":
        count += 1

    # Why: spaces around the text match whole words only, so " late " is found
    # in "my order is late" but NOT inside "chocolate"
    clean_text = message.replace("?", "").replace(".", "").replace(",", "").replace("!", "")
    padded = " " + clean_text + " "

    if message == "":
        print(f"{BOT_NAME}: Please type a question.")
    elif message == "bye":
        print(f"{BOT_NAME}: Goodbye, {name}!")
        break
    # Bonus: human handover ends the chat
    elif message == "agent" or message == "human":
        print(f"{BOT_NAME}: Connecting you to a human...")
        break
    # Why: complaints come first, so an unhappy customer always reaches a human
    elif " refund " in padded or " complaint " in padded or " late " in padded:
        complaints += 1
        answered += 1
        print(f"{BOT_NAME}: Sorry about that. A manager will call you within 1 hour.")
    elif message == "/quote":
        item = input(f"{BOT_NAME}: Which item? ").strip().lower()
        quantity_text = input(f"{BOT_NAME}: How many? ").strip()

        if item == "cupcake":
            unit_price = CUPCAKE_PRICE
        elif item == "brownie":
            unit_price = BROWNIE_PRICE
        elif item == "chocolate cake":
            unit_price = CAKE_PRICE
        else:
            unit_price = 0

        # Bonus: isdigit() stops letters from crashing int()
        if quantity_text.isdigit():
            quantity = int(quantity_text)
        else:
            quantity = 0

        if unit_price == 0:
            print(f"{BOT_NAME}: Sorry, {item} is not on our menu.")
        elif quantity < 1:
            print(f"{BOT_NAME}: Please type the quantity as a whole number, like 12.")
        else:
            subtotal = unit_price * quantity
            if quantity >= BULK_MIN_QTY:
                discount = subtotal * BULK_DISCOUNT
            else:
                discount = 0
            # Why: delivery is judged AFTER the discount, so the customer is charged fairly
            after_discount = subtotal - discount
            if after_discount >= FREE_DELIVERY_LIMIT:
                delivery = 0
            else:
                delivery = DELIVERY_FEE
            total = after_discount + delivery

            quotes += 1
            answered += 1
            quoted_value += total
            print(f"{BOT_NAME}: {quantity} x {item} = {subtotal:,} PKR")
            print(f"     Discount: -{discount:,.0f} PKR | Delivery: {delivery:,} PKR")
            print(f"     Total: {total:,.0f} PKR")
    elif "hour" in message or "open" in message or "timing" in message or " time " in padded:
        answered += 1
        print(f"{BOT_NAME}: We are open 9 am to 10 pm, every day.")
    elif "delivery" in message:
        answered += 1
        print(f"{BOT_NAME}: Delivery is {DELIVERY_FEE:,} PKR, free from {FREE_DELIVERY_LIMIT:,} PKR.")
    elif "price" in message or "cost" in message or "how much" in message:
        answered += 1
        print(f"{BOT_NAME}: Cupcake {CUPCAKE_PRICE:,} PKR, brownie {BROWNIE_PRICE:,} PKR, "
              f"chocolate cake {CAKE_PRICE:,} PKR. Type /quote for an exact price.")
    elif "where" in message or "location" in message:
        answered += 1
        print(f"{BOT_NAME}: We are on Main Boulevard, Lahore.")
    # Why: short words like "hi" use == / startswith, because "hi" hides inside "which"
    elif message == "hi" or message == "hello" or message.startswith("salam"):
        answered += 1
        print(f"{BOT_NAME}: Hello {name}! Ask me anything or type /quote.")
    else:
        unanswered += 1
        question = message.capitalize()[:150]
        tokens = round(len(question) / 4)

        # Why: private details never leave the computer, so they go to the local model at cost 0
        if "phone" in message or "cnic" in message or "my address" in message:
            model = "llama3.2 (local, private)"
            cost = 0
        else:
            model = "small cloud model (cheap)"
            cost = tokens / 1000 * CLOUD_PRICE_PER_1K
        ai_cost += cost

        prompt = f"""You are the assistant of {BUSINESS}.
Answer {name} in 2 short lines. Question: {question}"""
        print(f"{BOT_NAME}: Good question! Routed to: {model} (~{tokens} tokens)")
        print(f"     Prompt: {prompt}")

print("=" * 50)
print("Session report")
print(f"Messages: {count} | Answered by rules: {answered} | Unanswered: {unanswered}")
print(f"Quotes: {quotes} | Quoted value: {quoted_value:,.0f} PKR | Complaints: {complaints}")
print(f"Estimated AI cost for unanswered questions: {ai_cost:.4f} PKR")
print("=" * 50)
```
